### **Librerias**

```python
import cv2
import numpy as np
from matplotlib import pyplot as plt
```

### **Primer Imagen**

Cargar imagen

```python
imagen = cv2.imread("ave1.jpg")
grises = cv2.cvtColor(imagen, cv2.COLOR_BGR2GRAY)
```

Aplicar umbrales

```python
_, binario = cv2.threshold(grises, 120, 255, cv2.THRESH_BINARY)
_, binario_inv = cv2.threshold(grises, 120, 255, cv2.THRESH_BINARY_INV)
_, truncar = cv2.threshold(grises, 120, 255, cv2.THRESH_TRUNC)
_, tozero = cv2.threshold(grises, 120, 255, cv2.THRESH_TOZERO)
_, tozero_inv = cv2.threshold(grises, 120, 255, cv2.THRESH_TOZERO_INV)
```

Mostrar resultados

```python
titulos = ['Original', 'Binario', 'Binario Invertido', 'Truncado', 'A Cero', 'A Cero Invertido']
imagenes = [grises, binario, binario_inv, truncar, tozero, tozero_inv]

plt.figure(figsize=(10,6))
for i in range(6):
    plt.subplot(2,3,i+1)
    plt.imshow(imagenes[i], cmap='gray')
    plt.title(titulos[i])
    plt.xticks([]), plt.yticks([])

plt.tight_layout()
plt.show()
```
