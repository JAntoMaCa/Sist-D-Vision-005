IMPORTAR LIBRERÍAS

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

 FUNCIÓN PARA MOSTRAR IMÁGENES

```python
def mostrar_imagen(titulo, imagen):
    """
    Función personalizada para mostrar imágenes con Matplotlib
    Convierte automáticamente de BGR a RGB cuando es necesario
    """
    plt.figure(figsize=(8, 6))
    plt.title(titulo, fontsize=14, fontweight='bold')

    # Si la imagen tiene 3 canales (color), convertir BGR a RGB
    if len(imagen.shape) == 3:
        imagen_mostrar = cv2.cvtColor(imagen, cv2.COLOR_BGR2RGB)
        plt.imshow(imagen_mostrar)
    else:
        # Para imágenes en escala de grises
        plt.imshow(imagen, cmap='gray')

    plt.axis('on')
    plt.grid(False)
    plt.tight_layout()
    plt.show()
```

REAR FONDO NEGRO (IMAGEN BASE)

```python
print("IMAGEN BASE")
imagen = np.zeros((512, 512, 3), np.uint8)
mostrar_imagen("Fondo Negro - Imagen Base", imagen)
```

DIBUJAR LÍNEA

```python
print("\n LÍNEA")
imagen_linea = imagen.copy()
cv2.line(imagen_linea, (100, 100), (400, 100), (0, 0, 255), 5)  # Línea roja
mostrar_imagen("Línea Roja", imagen_linea)
```

DIBUJAR RECTÁNGULO

```python
print("\n RECTÁNGULO ")
imagen_rect = imagen.copy()
cv2.rectangle(imagen_rect, (150, 150), (350, 300), (0, 255, 0), 4)  # Rectángulo verde
mostrar_imagen("Rectángulo Verde", imagen_rect)
```

DIBUJAR CÍRCULO

```python
print("\n CÍRCULO")
imagen_circ = imagen.copy()
cv2.circle(imagen_circ, (256, 256), 100, (255, 0, 0), 3)  # Círculo azul
mostrar_imagen("Círculo Azul", imagen_circ)
```

DIBUJAR POLÍGONO

```python
print("\n POLÍGONO")
imagen_poly = imagen.copy()
pts = np.array([[100, 300], [200, 150], [300, 300], [250, 400], [150, 400]], np.int32)
pts = pts.reshape((-1, 1, 2))
cv2.polylines(imagen_poly, [pts], True, (255, 255, 0), 3)  # Polígono amarillo
mostrar_imagen("Polígono Amarillo", imagen_poly)
```

AGREGAR TEXTO

```python
print("\n TEXTO ")
imagen_texto = imagen.copy()
texto = "OpenCV - Figuras Geometricas"
cv2.putText(imagen_texto, texto, (50, 450), cv2.FONT_HERSHEY_SIMPLEX, 0.8, (255, 255, 255), 2)
mostrar_imagen("Texto en Imagen", imagen_texto)
```

DIBUJO CREATIVO FINAL

```python
print("\n DIBUJO FINAL ")
dibujo_final = imagen.copy()

# Casa - Rectángulo principal
cv2.rectangle(dibujo_final, (150, 250), (350, 450), (0, 200, 200), -1)  # Relleno

# Techo - Triángulo (polígono)
pts_techo = np.array([[150, 250], [250, 150], [350, 250]], np.int32)
pts_techo = pts_techo.reshape((-1, 1, 2))
cv2.fillPoly(dibujo_final, [pts_techo], (200, 100, 100))  # Relleno

# Puerta - Rectángulo
cv2.rectangle(dibujo_final, (220, 350), (280, 450), (100, 100, 200), -1)

# Ventanas - Círculos
cv2.circle(dibujo_final, (190, 300), 25, (200, 200, 100), -1)
cv2.circle(dibujo_final, (310, 300), 25, (200, 200, 100), -1)

# Sol - Círculo
cv2.circle(dibujo_final, (400, 100), 40, (0, 255, 255), -1)

# Rayos de sol - Líneas
for i in range(0, 360, 45):
    x1 = 400 + int(50 * np.cos(np.radians(i)))
    y1 = 100 + int(50 * np.sin(np.radians(i)))
    x2 = 400 + int(70 * np.cos(np.radians(i)))
    y2 = 100 + int(70 * np.sin(np.radians(i)))
    cv2.line(dibujo_final, (x1, y1), (x2, y2), (0, 255, 255), 3)

# Texto descriptivo
cv2.putText(dibujo_final, "Casa con Jardin - OpenCV", (100, 50),
            cv2.FONT_HERSHEY_COMPLEX, 0.7, (255, 255, 255), 2)

mostrar_imagen("Casa con Jardin - OpenCV", dibujo_final)

```
