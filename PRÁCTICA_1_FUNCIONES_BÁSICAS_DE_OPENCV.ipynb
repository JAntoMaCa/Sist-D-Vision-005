IMPORTAR LIBRERÍAS

```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
```

DEFINIR FUNCIÓN PARA MOSTRAR IMÁGENES

```python
def mostrar_imagen(titulo, img):
    """
    Función personalizada para mostrar imágenes con Matplotlib
    Convierte automáticamente de BGR a RGB cuando es necesario
    """
    plt.figure(figsize=(6, 4))
    plt.title(titulo, fontsize=14, fontweight='bold')

    # Si la imagen tiene 3 canales (color), convertir BGR a RGB
    if len(img.shape) == 3:
        img_mostrar = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
        plt.imshow(img_mostrar)
    else:
        # Para imágenes en escala de grises o binarias
        plt.imshow(img, cmap='gray')

    plt.axis('on')
    plt.grid(False)
    plt.tight_layout()
    plt.show()
```

CARGAR LA IMAGEN ORIGINAL

```python
print("CARGANDO IMAGEN :....")
imagen = cv2.imread("Foto (1).jpeg")
```

MOSTRAR IMAGEN ORIGINAL

```python
print("\n  IMAGEN ORIGINAL ")
mostrar_imagen("Imagen Original - Foto (1).jpeg", imagen)
```

ONVERTIR A ESCALA DE GRISES

```python
print("\nCONVERSIÓN A ESCALA DE GRISES ")
gris = cv2.cvtColor(imagen, cv2.COLOR_BGR2GRAY)
mostrar_imagen("Escala de Grises", gris)
```

TRABAJAR CON CANALES DE COLOR

```python
print("\n=== SEPARACIÓN DE CANALES DE COLOR ===")

# Separar los canales B, G, R
B, G, R = cv2.split(imagen)

print("🔵 Canal Azul (Blue)")
mostrar_imagen("Canal Azul (B)", B)

print("🟢 Canal Verde (Green)")
mostrar_imagen("Canal Verde (G)", G)

print("🔴 Canal Rojo (Red)")
mostrar_imagen("Canal Rojo (R)", R)
```

APLICAR FILTRO CANNY PARA DETECCIÓN DE BORDES

```python
umbral_bajo = 50
umbral_alto = 150

bordes = cv2.Canny(gris, umbral_bajo, umbral_alto)
mostrar_imagen(f"Detección de Bordes - Canny ({umbral_bajo}, {umbral_alto})", bordes)
```
