Librerías

```python
import cv2
from matplotlib import pyplot as plt
import numpy as np
```

Cálculo del histograma en escala de grises

```python
def mostrar_histograma_grises(ruta_imagen, nombre_imagen):

    # Leer la imagen en escala de grises
    img_gray = cv2.imread(ruta_imagen, 0)

    # Calcular histograma: frecuencia de píxeles entre 0 y 255
    hist = cv2.calcHist([img_gray], [0], None, [256], [0, 256])

    # Mostrar imagen y su histograma
    plt.figure(figsize=(10,5))

    # Imagen original
    plt.subplot(1,2,1)
    plt.imshow(img_gray, cmap='gray')
    plt.title(f"Imagen en escala de grises - {nombre_imagen}")
    plt.axis("off")

    # Histograma
    plt.subplot(1,2,2)
    plt.plot(hist, color='black')
    plt.title(f"Histograma de {nombre_imagen}")
    plt.xlabel("Intensidad de píxeles (0-255)")
    plt.ylabel("Frecuencia")
    plt.grid(True)

    plt.show()
```

Cálculo del histograma en color (RGB)

```python
def mostrar_histograma_rgb(ruta_imagen, nombre_imagen):

    # Leer imagen en formato BGR (por defecto en OpenCV)
    img_bgr = cv2.imread(ruta_imagen)
    # Convertir de BGR a RGB para que los colores se muestren correctamente
    img_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)

    # Crear figura
    plt.figure(figsize=(10,5))

    # Imagen original
    plt.subplot(1,2,1)
    plt.imshow(img_rgb)
    plt.title(f"Imagen original - {nombre_imagen}")
    plt.axis("off")

    # Graficar histogramas de los tres canales (Rojo, Verde, Azul)
    plt.subplot(1,2,2)
    colores = ('r', 'g', 'b')
    for i, col in enumerate(colores):
        hist = cv2.calcHist([img_bgr], [i], None, [256], [0,256])
        plt.plot(hist, color=col)
        plt.xlim([0,256])
    plt.title(f"Histograma RGB - {nombre_imagen}")
    plt.xlabel("Intensidad de píxeles")
    plt.ylabel("Frecuencia")
    plt.grid(True)

    plt.show()
```

Resultados

```python
# Lista de imágenes con sus nombres descriptivos
imagenes = [
    ("ave1 (1).jpg", "Ave 1"),
    ("ave2 (1).jpg", "Ave 2"),
    ("ave3 (2).jpg", "Ave 3"),
    ("Foto.jpeg", "Foto de portada")
]

# Generar histogramas para cada imagen
for ruta, nombre in imagenes:
    print(f"\n===============================")
    print(f"Procesando imagen: {nombre}")
    print(f"===============================")
    mostrar_histograma_grises(ruta, nombre)
    mostrar_histograma_rgb(ruta, nombre)
```
